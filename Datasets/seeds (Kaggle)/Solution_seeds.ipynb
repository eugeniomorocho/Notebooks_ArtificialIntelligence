```python
import pandas as pd
X_train_seeds = pd.read_csv('/Users/eugenio/Documents/Notebooks_ArtificialIntelligence/Datasets/seeds (Kaggle)/X_train_seeds.csv')
y_train_seeds = pd.read_csv('/Users/eugenio/Documents/Notebooks_ArtificialIntelligence/Datasets/seeds (Kaggle)/y_train_seeds.csv')
X_test_seeds = pd.read_csv('/Users/eugenio/Documents/Notebooks_ArtificialIntelligence/Datasets/seeds (Kaggle)/X_test_seeds.csv')
sample_submission_seeds = pd.read_csv('/Users/eugenio/Documents/Notebooks_ArtificialIntelligence/Datasets/seeds (Kaggle)/sample_submission_seeds.csv')
```

```python
# Scale the features using StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_seeds)
X_test_scaled = scaler.transform(X_test_seeds)
```

```python
y_train_seeds.value_counts()
```

```python
# Fit a MLPClassifier model
from sklearn.neural_network import MLPClassifier
mlp = MLPClassifier(hidden_layer_sizes=(50, 25), max_iter=500, random_state=42)
mlp.fit(X_train_scaled, y_train_seeds.values)
```

```python
# Fit a MLPClassifier model with GridSearchCV
from sklearn.model_selection import GridSearchCV
param_grid = {
    'hidden_layer_sizes': [(50, 25), (100, 50), (150, 75), (200, 100, 50), (100, 100, 50, 25)],
    'activation': ['relu', 'tanh'],
    'solver': ['adam', 'sgd'],
    'alpha': [0.0001, 0.001, 0.01]
}
mlp = MLPClassifier(max_iter=500, random_state=42)
grid_search = GridSearchCV(estimator=mlp, param_grid=param_grid, cv=10, scoring='accuracy')
grid_search.fit(X_train_scaled, y_train_seeds)
best_mlp = grid_search.best_estimator_
best_mlp.fit(X_train_scaled, y_train_seeds)
# Make predictions on the test set
y_pred = best_mlp.predict(X_test_scaled)
y_pred
```

```python
# Print the best hyperparameters and the corresponding accuracy
print("Best Hyperparameters:", grid_search.best_params_)
print("Best Cross-Validation Accuracy:", grid_search.best_score_)
```

```python
# Fit a MLPClassifier model with GridSearchCV with a nested parameter grid to search for different numbers of layers and neurons per layer
from sklearn.model_selection import GridSearchCV

# Define a range of number of hidden layers from 1 to 3 and neurons per layer from 5 to 100 with step 5 
param_grid = {
    'hidden_layer_sizes': [(n,) for n in range(5, 105, 5)] + 
                          [(n, m) for n in range(5, 105, 5) for m in range(5, 105, 5)] + 
                          [(n, m, o) for n in range(5, 105, 5) for m in range(5, 105, 5) for o in range(5, 105, 5)],
    'activation': ['relu', 'tanh'],
    'solver': ['adam', 'sgd'],
    'alpha': [0.0001, 0.001, 0.01]
}

# Use EarlyStopping to prevent overfitting
mlp = MLPClassifier(max_iter=500, random_state=42, early_stopping=True)
grid_search = GridSearchCV(estimator=mlp, param_grid=param_grid, cv=10, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train_scaled, y_train_seeds)
best_mlp = grid_search.best_estimator_
best_mlp.fit(X_train_scaled, y_train_seeds)
# Make predictions on the test set
y_pred = best_mlp.predict(X_test_scaled)
y_pred
```
