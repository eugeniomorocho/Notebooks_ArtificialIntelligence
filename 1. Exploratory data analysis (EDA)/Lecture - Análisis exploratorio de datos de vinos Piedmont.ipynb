## Análisis exploratorio de datos de vinos Piedmont

### **Introducción:**

En esta lectura, realizaremos un análisis exploratorio de datos del conjunto de datos de vinos. Este conjunto de datos contiene características químicas de muestras de vinos, clasificadas en tres clases diferentes. Utilizaremos herramientas de visualización y análisis estadístico para comprender mejor la estructura y las relaciones en los datos.


El conjunto de datos de vino (wine dataset) es un conjunto de datos clásico en el campo del data mining. Contiene 178 muestras de vino cultivadas en la [región Piedmont](https://en.wikipedia.org/wiki/Piedmont) de Italia pero derivadas de tres cultivares diferentes (Barolo, Grignolio and Barbera). Cada muestra de vino tiene 13 características químicas medidas, que son las variables del conjunto de datos. Aquí están las variables:

1. **Alcohol**: la cantidad de alcohol en el vino.
2. **Malic acid**: un ácido orgánico que se encuentra en los vinos.
3. **Ash**: una medida de la cantidad total de minerales presentes en el vino.
4. **Alcalinity of ash**: una medida de la alcalinidad de los minerales presentes en el vino.
5. **Magnesium**: la cantidad de magnesio en el vino.
6. **Total phenols**: una medida de los compuestos fenólicos presentes en el vino.
7. **Flavanoids**: una clase de fenoles que se encuentran en los vinos.
8. **Nonflavanoid phenols**: una medida de los fenoles no flavonoides presentes en el vino.
9. **Proanthocyanins**: una clase de fenoles que se encuentran en los vinos.
10. **Color intensity**: una medida de la intensidad del color del vino.
11. **Hue**: una medida del color del vino.
12. **OD280/OD315 of diluted wines**: una medida de la densidad óptica del vino.
13. **Proline**: una medida de la cantidad de prolina en el vino, un aminoácido que puede influir en el sabor del vino.

Cada una de estas variables puede influir en la calidad y el sabor del vino, y pueden ser útiles para predecir el cultivar del vino.

Fuente: https://archive.ics.uci.edu/ml/datasets/Wine

#### Distribución de clases

Este código genera un gráfico de barras que muestra la distribución de clases en el DataFrame `wine_df`. 

La función `sns.countplot()` de la biblioteca seaborn se utiliza para crear el gráfico de barras. El argumento `x='Class'` especifica que las barras del gráfico deben representar las diferentes clases en la columna 'Class' del DataFrame. El argumento `data=wine_df` especifica que los datos para el gráfico provienen del DataFrame `wine_df`. El argumento `palette='Set2'` especifica que se debe usar la paleta de colores 'Set2' para las barras del gráfico.

Las funciones `plt.title()`, `plt.xlabel()`, y `plt.ylabel()` se utilizan para establecer el título del gráfico y las etiquetas de los ejes x e y, respectivamente.

Finalmente, `plt.show()` se utiliza para mostrar el gráfico.

```python
%pip install seaborn
%pip install matplotlib

# Visualizar la distribución de clases
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de la visualización
sns.countplot(x='Class', data=wine_df, palette='Set2')

# Configuramos el título y los ejes
plt.title('Distribución de Clases')
plt.xlabel('Clase')
plt.ylabel('Cantidad')
plt.show()
```

#### Matriz de correlación entre características

La matriz de correlación te permite ver las correlaciones entre las diferentes características en el conjunto de datos `wine_df`. Los valores en la matriz de correlación varían de -1 a 1, donde:

- Un valor cercano a 1 indica una fuerte correlación positiva: si una característica aumenta, la otra también tiende a aumentar.
- Un valor cercano a -1 indica una fuerte correlación negativa: si una característica aumenta, la otra tiende a disminuir.
- Un valor cercano a 0 indica que no hay una correlación lineal entre las dos características.

Primero, se calcula la matriz de correlación utilizando el método `corr()` del DataFrame. Esta matriz de correlación es un DataFrame donde el valor en la i-ésima fila y la j-ésima columna es la correlación entre la i-ésima y la j-ésima columna del DataFrame original.

Luego, se crea una figura con un tamaño específico utilizando `plt.figure(figsize=(10, 8))`.

Después, se utiliza la función `sns.heatmap()` de la biblioteca seaborn para visualizar la matriz de correlación. Los argumentos `annot=True` y `cmap='coolwarm'` especifican que los valores de correlación deben ser anotados en el heatmap y que se debe usar el mapa de colores 'coolwarm', respectivamente.

Finalmente, se establece el título del gráfico con `plt.title('Matriz de Correlación')` y se muestra el gráfico con `plt.show()`.

```python
# Calcular matriz de correlación
correlation_matrix = wine_df.corr()
print(correlation_matrix)
```

```python
# Visualizar matriz de correlación
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Matriz de Correlación')
plt.show()
```

Al observar la matriz de correlación, puedes inferir las relaciones entre las diferentes características del vino. Por ejemplo, se puede encontrar una fuerte correlación positiva entre la cantidad de flavonoides y la cantidad de fenoles totales, esto indica que los vinos que tienen altos niveles de fenoles totales también tienden a tener altos niveles de flavonoides.

Por otro lado, se encuentra una correlación negativa entre la alcalinidad de la ceniza y la cantidad de alcohol, esto podría indicar que los vinos con alta alcalinidad de ceniza tienden a tener una menor cantidad de alcohol.

Estas inferencias pueden ser útiles para entender las características del vino y también pueden ser útiles para la construcción y la interpretación de modelos de data mining.

#### Relación entre dos características

Generamos un gráfico de dispersión utilizando la biblioteca seaborn. El gráfico de dispersión muestra la relación entre las variables 'Alcohol' y 'Flavanoides' en el DataFrame `wine_df`.

La función `sns.scatterplot()` se utiliza para crear el gráfico de dispersión. Los argumentos `x='Alcohol'` y `y='Flavanoides'` especifican las variables que se utilizarán en los ejes x e y del gráfico, respectivamente.

El argumento `hue='Class'` significa que los puntos en el gráfico se colorearán de acuerdo con la columna 'Class' en el DataFrame. Esto puede ayudarte a visualizar cómo se distribuyen las diferentes clases en relación con las variables 'Alcohol' y 'Flavanoides'.

Las funciones `plt.title()`, `plt.xlabel()` y `plt.ylabel()` se utilizan para establecer el título del gráfico y las etiquetas de los ejes x e y, respectivamente.

Finalmente, `plt.show()` se utiliza para mostrar el gráfico.

```python
# Relación entre 'Alcohol' y 'Flavanoids'
sns.scatterplot(x='Alcohol', y='Flavanoids', hue='Class', data=wine_df)
plt.title('Relación entre Alcohol y Flavanoides')
plt.xlabel('Alcohol')
plt.ylabel('Flavanoides')
plt.show()
```

En este gráfico se muestra la relación entre las características 'alcohol' y 'flavanoids' como un ejemplo específico de cómo explorar la relación entre dos variables continuas en el conjunto de datos de vinos. Estas dos características son conocidas por su influencia en la calidad y el sabor del vino, y su relación puede ser de interés para los expertos en vinos.
