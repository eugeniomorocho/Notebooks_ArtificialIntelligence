## 5. Tree-Based Models: Bagging and Random Forests

Tree-based models are supervised learning algorithms that split the feature space into homogeneous regions to predict a target variable. A single decision tree is easy to interpret, but it is often prone to overfitting.

**Bagging** (Bootstrap Aggregating) improves stability by creating multiple bootstrap samples from the training data, fitting one tree per sample, and combining predictions:
- Regression: average of predictions
- Classification: majority vote

This process mainly reduces model variance.

**Random Forests** extend bagging by adding extra randomness at each split: instead of evaluating all features, each node considers only a random subset of features. This increases tree diversity, reduces correlation among trees, and usually improves generalization while keeping overfitting under control.

```python
# Example in Python: Bagging vs Random Forest on Iris
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load Iris dataset
X, y = load_iris(return_X_y=True)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# Bagging with decision trees
bagging = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=42),
    n_estimators=100,
    random_state=42
)
bagging.fit(X_train, y_train)
bagging_pred = bagging.predict(X_test)

# Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print(f"Bagging accuracy: {accuracy_score(y_test, bagging_pred):.3f}")
print(f"Random Forest accuracy: {accuracy_score(y_test, rf_pred):.3f}")
print(f"Random Forest feature importances: {rf.feature_importances_}")
```

Each tree sees a different bootstrap sample, and in Random Forest each split also sees only a subset of features. This makes predictions more robust and less sensitive to noise than a single tree.

## 6. Tree Ensemble Methods: Boosting

Boosting trains weak learners sequentially, where each new model focuses on correcting errors made by previous ones.

- **AdaBoost** increases the weight of misclassified observations at each iteration.
- **Gradient Boosting** fits each new learner to the residual errors (or gradients of the loss), directly optimizing the objective function.

Compared with bagging, boosting often achieves higher predictive performance, but it can be more sensitive to hyperparameters and noisy data.

```python
# Example in Python: AdaBoost and Gradient Boosting on Iris
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score

# Load Iris dataset
X, y = load_iris(return_X_y=True)

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# AdaBoost
adb = AdaBoostClassifier(n_estimators=50, random_state=42)
adb.fit(X_train, y_train)
adb_pred = adb.predict(X_test)

# Gradient Boosting (added as an additional key boosting model)
gb = GradientBoostingClassifier(n_estimators=100, random_state=42)
gb.fit(X_train, y_train)
gb_pred = gb.predict(X_test)

print(f"AdaBoost accuracy: {accuracy_score(y_test, adb_pred):.3f}")
print(f"Gradient Boosting accuracy: {accuracy_score(y_test, gb_pred):.3f}")
```

AdaBoost gives more weight to hard-to-classify examples at each iteration, combining weak learners into a stronger final model. Gradient Boosting generalizes this idea by optimizing a differentiable loss function and is a direct foundation for modern methods such as XGBoost and LightGBM.

## Conclusions

The supervised models reviewed provide a strong foundation for applied data science.

- **Linear Regression** is a natural starting point for continuous targets due to interpretability and efficiency, but it relies on assumptions that may fail in complex settings.
- **Logistic Regression** extends the framework to categorical targets and probabilistic interpretation (including multiclass settings), though linear decision boundaries can limit performance.
- **Naive Bayes** is simple and fast, especially useful in high-dimensional tasks such as text classification, but its conditional independence assumption is often unrealistic.
- **k-NN** is intuitive and flexible for non-linear patterns, yet it can struggle with scalability and high-dimensional data.
- **Decision Trees** are interpretable but prone to overfitting.
- **Bagging and Random Forests** improve robustness by reducing variance and adding diversity across trees, often delivering strong accuracy and feature-importance insights.
- **Boosting methods** (AdaBoost, Gradient Boosting) further improve performance by correcting errors sequentially, at the cost of more careful tuning and computational effort.

Together, these methods are essential to understand supervised learning fundamentals and prepare for more advanced techniques such as XGBoost, LightGBM, hybrid ensembles, and deep learning.
