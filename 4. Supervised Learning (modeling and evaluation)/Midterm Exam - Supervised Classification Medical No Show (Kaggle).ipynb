# 🧠 Mini AI Competition: "Will the Patient Show Up?"

## 👥 Format

- Teams of 2 students
- Collaborative work with live support from the instructor
- This is a formative activity with academic value (no numerical grade)


## 📊 Dataset

- **Title**: Medical Appointment No Shows  
- **Source**: [Kaggle Dataset](https://www.kaggle.com/datasets/joniarroba/noshowappointments)  
- **Size**: Over 110,000 records  
- **Goal**: Predict whether a patient will show up for their medical appointment (`No-show` column)
