# Tree-Based Models and Ensemble Methods
### Simple Tutorial ("like explaining with apples")

This notebook introduces Decision Trees, Bagging, Random Forest, and Boosting methods in a simple and intuitive way.

## 1. Decision Trees

Imagine choosing a fruit:
- Is it red?
- Is it sweet?

A Decision Tree works like this: asking questions step by step.

Train

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True)

#Split the data into training and testing sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
print('Accuracy (train set):', model.score(X_train, y_train))
print('Accuracy (test set):', model.score(X_test, y_test))
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier

model = BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=50)
model.fit(X_train, y_train)

print('Accuracy (train set):', model.score(X_train, y_train))
print('Accuracy (test set):', model.score(X_test, y_test))
```

## 3. Random Forest

Each tree sees:
- Different data
- Different features

This makes the model stronger and more robust.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=0)
rf.fit(X, y)
print('Accuracy:', rf.score(X, y))
```

## 4. Extra Trees

Even more randomness → often faster and stable.

```python
from sklearn.ensemble import ExtraTreesClassifier

model = ExtraTreesClassifier(n_estimators=100)
model.fit(X, y)
print('Accuracy:', model.score(X, y))
```

## 5. Boosting

Models learn from mistakes step by step.

Like studying harder the questions you got wrong.

## 6. AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier

adb = AdaBoostClassifier(n_estimators=50)
adb.fit(X, y)
print('Accuracy:', adb.score(X, y))
```

## 7. Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(n_estimators=100)
gb.fit(X, y)
print('Accuracy:', gb.score(X, y))
```

## Final Idea

If one model is good → many models are better.

If they learn from mistakes → even better.
