# Accuracy, Precision, Recall y F1-score

## 1. Matriz de confusión

|                | Predicción Negativa | Predicción Positiva |
|----------------|---------------------|---------------------|
| **Real Negativo** | TN (True Negative)  | FP (False Positive) |
| **Real Positivo** | FN (False Negative) | TP (True Positive)  |
---

## 2. Definiciones

### Accuracy (Exactitud)

$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$

- Mide el porcentaje total de aciertos.  
- Puede ser engañosa si las clases están **desbalanceadas**.  
- Ejemplo: si el 99% de los pacientes están sanos y el modelo siempre predice "sano", tendrá 99% de accuracy aunque no detecte a ningún enfermo.

---

### Precision (Precisión)

$Precision = \frac{TP}{TP + FP}$

- De todos los que predije como **positivos**, ¿cuántos realmente lo eran?  
- Importa cuando los **falsos positivos son costosos**.  
- Ejemplo: detección de intrusos → demasiadas alarmas falsas cansan al usuario.

---

### Recall (Sensibilidad / Exhaustividad)

$Recall = \frac{TP}{TP + FN}$

- De todos los que eran realmente **positivos**, ¿cuántos encontré?  
- Importa cuando los **falsos negativos son costosos**.  
- Ejemplo: diagnóstico médico → mejor detectar a todos los enfermos, aunque existan falsos positivos.

---

### F1-score

$F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}$

- Es la **media armónica** entre precision y recall.  
- Útil cuando las clases están **desbalanceadas** y se necesita equilibrio.  
- Ejemplo: clasificación de spam → queremos detectar la mayoría de spam (recall) sin marcar correos legítimos como spam (precision).

---

## 3. Analogía: Detector de incendios

- **Recall alto**: detecta todos los fuegos (pero puede sonar cuando cocinas → FP).  
- **Precision alta**: cada vez que suena, es realmente fuego (pero puede dejar escapar alguno → FN).  
- **F1-score**: equilibrio entre sonar demasiado y no sonar cuando debe.  
- **Accuracy**: mide aciertos globales, pero puede ser engañoso si casi nunca hay incendios.

---

## 4. ¿Cuándo usar cada métrica?

- **Accuracy** → cuando las clases están balanceadas.  
- **Precision** → cuando los **falsos positivos son graves** (ej: intrusos).  
- **Recall** → cuando los **falsos negativos son graves** (ej: medicina).  
- **F1-score** → cuando necesitas **equilibrio** y el dataset está desbalanceado.
