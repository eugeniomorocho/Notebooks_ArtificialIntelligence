### **Tutorial:** Using Meta's LLaMA Model

1. **Sign Up for Access**:
   - Visit the [Meta's LLaMA website](https://www.llama.com) and sign up for access to the LLaMA model. You may need to fill out a form and provide some details about your intended use case.

2. **Unique custom URL**:
   - Once your access is approved, you will receive an **unique custom URL** to your registered e-mail. This unique custom URL is essential for downloading the LLaMA model.

   - Save the unique custom URL, it will remain valid for 48 hours to download the model up to 5 times, and requests can be submitted multiple times.

3. **Install the LLaMA command line interface (CLI)**

    In your preferred environment run the command below: 

    `pip install llama-stack`

    Use -U option to update llama-stack if a previous version is already installed:

    `pip install llama-stack -U`

4. **List the available LLaMA models**

    See latest available models by running the following command and determine the model ID you wish to download:

    `llama model list`

5. **Select a LLaMA model to download**

    Select a desired model by running:

    `llama model download --source meta --model-id  MODEL_ID`

    replacing *MODEL_ID* with the desired model, for example *Llama3.2-1B*

6. **Specify custom URL**

    When the script asks for your **unique custom URL**, please paste the URL received via e-mail and start the download.

    The model will be downloaded to an address similar to: /Users/eugenio/.llama/checkpoints/Llama3.2-1B

**Note:** To remove the dowloaded models you can use:  
    - `llama model remove --model-id <MODEL_ID>`  
    - `llama cache clear`
